```python
import numpy as np
from datasets import load_dataset
import pandas as pd
from sklearn.model_selection import train_test_split
import os
dataset = load_dataset("FIdo-AI/ua-news")
```

```python
for item in dataset["train"]:
    title, text, tags, target = item["title"], item["text"], item["tags"], item["target"]
    print("Title: " + title)
    print("Text: " + text)
    print("Tags: " + tags)
    print("Target: " + target)
    break
```

```python
np.mean([len(i.split(' ')) for i in dataset["train"]['text']])
```

```python
unique_labels = set(dataset['train']['target'])
unique_labels
```

```python
df = pd.DataFrame(dataset['train'])
test_df = pd.DataFrame(dataset['test'])
```

```python
train_df, eval_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df['target'],
    random_state=42
)
```

```python
os.makedirs("ua-news", exist_ok=True)
train_df.to_csv("ua-news/train.csv", index=False)
eval_df.to_csv("ua-news/eval.csv", index=False)
test_df.to_csv("ua-news/test.csv", index=False)
```
